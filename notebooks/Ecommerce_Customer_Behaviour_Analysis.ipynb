```python
import pandas as pd

df = pd.read_excel('/content/Online Retail.xlsx')

print("Dataset loaded successfully!")
print("Shape:", df.shape)

df.head()
```

```python
print("Column Names:")
print(df.columns)

print("\nDataset Information:")
df.info()

print("\nFirst 5 Records:")
display(df.head())
```

```python
print("Missing Values:")
print(df.isnull().sum())

print("\nTotal Missing Values:")
print(df.isnull().sum().sum())

print("\nDuplicate Records:")
print(df.duplicated().sum())
```

```python
print("Cancelled Invoices:")
print(df['InvoiceNo'].astype(str).str.startswith('C').sum())

print("\nNegative Quantities:")
print((df['Quantity'] < 0).sum())

print("\nZero or Negative Prices:")
print((df['UnitPrice'] <= 0).sum())

print("\nMissing Descriptions:")
print(df['Description'].isnull().sum())

print("\nUnique Countries:")
print(df['Country'].nunique())
```

```python
print("Cancelled Invoice Examples:")
display(df[df['InvoiceNo'].astype(str).str.startswith('C')].head())

print("Negative Quantity Examples:")
display(df[df['Quantity'] < 0].head())

print("Zero or Negative Price Examples:")
display(df[df['UnitPrice'] <= 0].head())
```

```python
cleaned_df = df.drop_duplicates().copy()

cleaned_df = cleaned_df[
    ~cleaned_df['InvoiceNo'].astype(str).str.startswith('C')
]

cleaned_df = cleaned_df[
    (cleaned_df['Quantity'] > 0) &
    (cleaned_df['UnitPrice'] > 0)
]

cleaned_df = cleaned_df.dropna(subset=['Description'])

cleaned_df['TotalAmount'] = (
    cleaned_df['Quantity'] * cleaned_df['UnitPrice']
)

print("Original Dataset Shape:", df.shape)
print("Cleaned Dataset Shape:", cleaned_df.shape)

print("\nRemaining Missing Values:")
print(cleaned_df.isnull().sum())

print("\nRemaining Duplicates:")
print(cleaned_df.duplicated().sum())

print("\nCleaned Dataset:")
display(cleaned_df.head())
```

```python
cleaned_df.to_csv('/content/cleaned_online_retail.csv', index=False)

print("Cleaned dataset saved successfully!")
```

```python
df = pd.read_csv(
    '/content/cleaned_online_retail.csv',
    dtype={'InvoiceNo': str},
    low_memory=False
)

df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'])

df['Year'] = df['InvoiceDate'].dt.year
df['Month'] = df['InvoiceDate'].dt.month
df['MonthName'] = df['InvoiceDate'].dt.strftime('%B')

print("Dataset loaded without mixed-type warning!")
print(df.dtypes)
print("\nShape:", df.shape)
```

```python
print("Total Revenue:", round(df['TotalAmount'].sum(), 2))

print("Total Orders:", df['InvoiceNo'].nunique())

print("Total Customers:", df['CustomerID'].nunique())

print("Total Products:", df['StockCode'].nunique())

print("Total Countries:", df['Country'].nunique())

print("Average Order Value:",
      round(df.groupby('InvoiceNo')['TotalAmount'].sum().mean(), 2))
```

```python
print("Descriptive Statistics:")
display(df[['Quantity', 'UnitPrice', 'TotalAmount']].describe())

print("\nTop 10 Selling Products:")
display(
    df.groupby('Description')['Quantity']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

print("\nTop 10 Countries by Revenue:")
display(
    df.groupby('Country')['TotalAmount']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)
```

```python
print("Correlation Matrix:")
display(
    df[['Quantity', 'UnitPrice', 'TotalAmount']].corr()
)

q1 = df['Quantity'].quantile(0.25)
q3 = df['Quantity'].quantile(0.75)
iqr = q3 - q1

lower_limit = q1 - 1.5 * iqr
upper_limit = q3 + 1.5 * iqr

outliers = df[
    (df['Quantity'] < lower_limit) |
    (df['Quantity'] > upper_limit)
]

print("\nQuantity Outlier Analysis")
print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)
print("Number of Outliers:", len(outliers))

print("\nLargest Quantity Transactions:")
display(
    df.nlargest(5, 'Quantity')[
        ['InvoiceNo', 'Description', 'Quantity', 'UnitPrice', 'TotalAmount']
    ]
)
```

```python
import matplotlib.pyplot as plt

monthly_sales = df.groupby('Month')['TotalAmount'].sum()

plt.figure(figsize=(10, 5))
plt.plot(monthly_sales.index, monthly_sales.values, marker='o')

plt.title('Monthly Revenue Trend')
plt.xlabel('Month')
plt.ylabel('Revenue (£)')
plt.xticks(range(1, 13))
plt.grid(True)

plt.show()
```

```python
top_products = (
    df.groupby('Description')['Quantity']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(12, 6))
top_products.sort_values().plot(kind='barh')

plt.title('Top 10 Best-Selling Products')
plt.xlabel('Quantity Sold')
plt.ylabel('Product Name')
plt.tight_layout()

plt.show()
```

```python
country_sales = (
    df.groupby('Country')['TotalAmount']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(12, 6))
country_sales.sort_values().plot(kind='barh')

plt.title('Top 10 Countries by Revenue')
plt.xlabel('Revenue (£)')
plt.ylabel('Country')
plt.tight_layout()

plt.show()
```

```python
monthly_orders = df.groupby('Month')['InvoiceNo'].nunique()

plt.figure(figsize=(10, 5))
plt.bar(monthly_orders.index, monthly_orders.values)

plt.title('Monthly Order Count')
plt.xlabel('Month')
plt.ylabel('Number of Orders')
plt.xticks(range(1, 13))
plt.grid(axis='y')

plt.show()
```

```python
top_customers = (
    df.dropna(subset=['CustomerID'])
    .groupby('CustomerID')['TotalAmount']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(12, 6))
top_customers.sort_values().plot(kind='barh')

plt.title('Top 10 Customers by Revenue')
plt.xlabel('Revenue (£)')
plt.ylabel('Customer ID')
plt.tight_layout()

plt.show()
```

```python
import os
import matplotlib.pyplot as plt

os.makedirs('/content/visualizations', exist_ok=True)

df['YearMonth'] = df['InvoiceDate'].dt.to_period('M').astype(str)

monthly_sales = df.groupby('YearMonth')['TotalAmount'].sum()

plt.figure(figsize=(12, 5))
plt.plot(monthly_sales.index, monthly_sales.values, marker='o')
plt.title('Monthly Revenue Trend')
plt.xlabel('Month')
plt.ylabel('Revenue (£)')
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.savefig('/content/visualizations/monthly_revenue.png')
plt.show()

top_products = df.groupby('Description')['Quantity'].sum().nlargest(10)

plt.figure(figsize=(12, 6))
top_products.sort_values().plot(kind='barh')
plt.title('Top 10 Best-Selling Products')
plt.xlabel('Quantity Sold')
plt.ylabel('Product Name')
plt.tight_layout()
plt.savefig('/content/visualizations/top_products.png')
plt.show()

country_sales = df.groupby('Country')['TotalAmount'].sum().nlargest(10)

plt.figure(figsize=(12, 6))
country_sales.sort_values().plot(kind='barh')
plt.title('Top 10 Countries by Revenue')
plt.xlabel('Revenue (£)')
plt.ylabel('Country')
plt.tight_layout()
plt.savefig('/content/visualizations/country_revenue.png')
plt.show()

monthly_orders = df.groupby('YearMonth')['InvoiceNo'].nunique()

plt.figure(figsize=(12, 5))
plt.bar(monthly_orders.index, monthly_orders.values)
plt.title('Monthly Order Count')
plt.xlabel('Month')
plt.ylabel('Number of Orders')
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig('/content/visualizations/monthly_orders.png')
plt.show()

top_customers = (
    df.dropna(subset=['CustomerID'])
    .groupby('CustomerID')['TotalAmount']
    .sum()
    .nlargest(10)
)

plt.figure(figsize=(12, 6))
top_customers.sort_values().plot(kind='barh')
plt.title('Top 10 Customers by Revenue')
plt.xlabel('Revenue (£)')
plt.ylabel('Customer ID')
plt.tight_layout()
plt.savefig('/content/visualizations/top_customers.png')
plt.show()

print("All five visualizations saved successfully!")
```
